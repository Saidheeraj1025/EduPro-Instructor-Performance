```python
import pandas as pd

# Ensure the required datasets exist before using them
required_vars = ["transactions", "teachers", "courses"]
missing_vars = [var for var in required_vars if var not in globals()]

if missing_vars:
    file_path = globals().get("file_path", "EduPro Online Platform.xlsx")
    if "transactions" not in globals():
        transactions = pd.read_excel(file_path, sheet_name="Transactions")
    if "teachers" not in globals():
        teachers = pd.read_excel(file_path, sheet_name="Teachers")
    if "courses" not in globals():
        courses = pd.read_excel(file_path, sheet_name="Courses")

# Ensure the merged dataframe exists before using it
if "df" not in globals():
    df = transactions.merge(teachers, on="TeacherID", how="left")
    df = df.merge(courses, on="CourseID", how="left")

# In this dataset, the merged df already has a single CourseRating column.
# Some earlier merge versions produced CourseRating_x, so use whichever exists.
course_rating_col = (
    "CourseRating_x" if "CourseRating_x" in df.columns else "CourseRating"
)

experience_course_corr = df[
    ["YearsOfExperience", course_rating_col]
].corr().iloc[0, 1]

print(
    "Experience vs Course Rating correlation:",
    round(experience_course_corr, 3)
)
```

```python
teacher_course_corr = df[
    ["TeacherRating", "CourseRating"]
].corr().iloc[0, 1]

print(
    "Teacher Rating vs Course Rating correlation:",
    round(teacher_course_corr, 3)
)
```

```python
# Ensure the aggregated instructor performance table exists
if "instructor_performance" not in globals():
    if "df" not in globals():
        df = transactions.merge(teachers, on="TeacherID", how="left")
        df = df.merge(courses, on="CourseID", how="left")

    instructor_performance = (
        df.groupby(
            [
                "TeacherID",
                "TeacherName",
                "Gender",
                "Expertise",
                "YearsOfExperience",
                "TeacherRating"
            ]
        )
        .agg(
            CourseCount=("CourseID", "nunique"),
            EnrollmentCount=("TransactionID", "count"),
            AverageCourseRating=("CourseRating", "mean")
        )
        .reset_index()
    )

def rating_tier(rating):
    if rating >= 4:
        return "High"
    elif rating >= 3:
        return "Medium"
    else:
        return "Low"


instructor_performance["RatingTier"] = (
    instructor_performance["TeacherRating"]
    .apply(rating_tier)
)

display(
    instructor_performance[
        [
            "TeacherName",
            "TeacherRating",
            "RatingTier"
        ]
    ].head(10)
)
```

```python
tier_enrollment = (
    instructor_performance
    .groupby("RatingTier")
    ["EnrollmentCount"]
    .sum()
    .sort_values(ascending=False)
)

print(tier_enrollment)
```

```python
plt.figure(figsize=(8, 5))

tier_enrollment.plot(kind="bar")

plt.title("Enrollment Volume by Instructor Rating Tier")
plt.xlabel("Instructor Rating Tier")
plt.ylabel("Enrollment Count")

plt.tight_layout()
plt.show()
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

instructor_performance[
    ["TeacherRating_Score",
     "CourseRating_Score",
     "Enrollment_Score"]
] = scaler.fit_transform(
    instructor_performance[
        [
            "TeacherRating",
            "AverageCourseRating",
            "EnrollmentCount"
        ]
    ]
)

display(instructor_performance.head())
```

```python
instructor_performance["InstructorPerformanceScore"] = (
    instructor_performance["TeacherRating_Score"] * 0.40 +
    instructor_performance["CourseRating_Score"] * 0.40 +
    instructor_performance["Enrollment_Score"] * 0.20
)

instructor_performance["InstructorPerformanceScore"] = (
    instructor_performance["InstructorPerformanceScore"] * 100
)

display(
    instructor_performance[
        [
            "TeacherID",
            "TeacherName",
            "Expertise",
            "YearsOfExperience",
            "TeacherRating",
            "AverageCourseRating",
            "EnrollmentCount",
            "InstructorPerformanceScore"
        ]
    ].sort_values(
        "InstructorPerformanceScore",
        ascending=False
    ).head(10)
)
```

```python
def performance_tier(score):
    if score >= 75:
        return "High Performer"
    elif score >= 50:
        return "Medium Performer"
    else:
        return "Low Performer"


instructor_performance["PerformanceTier"] = (
    instructor_performance[
        "InstructorPerformanceScore"
    ].apply(performance_tier)
)

display(
    instructor_performance[
        [
            "TeacherName",
            "InstructorPerformanceScore",
            "PerformanceTier"
        ]
    ].sort_values(
        "InstructorPerformanceScore",
        ascending=False
    )
)
```

```python
top_10_instructors = (
    instructor_performance
    .sort_values(
        "InstructorPerformanceScore",
        ascending=False
    )
    .head(10)
)

plt.figure(figsize=(12, 6))

sns.barplot(
    data=top_10_instructors,
    x="InstructorPerformanceScore",
    y="TeacherName"
)

plt.title("Top 10 Instructor Performance Leaderboard")
plt.xlabel("Performance Score")
plt.ylabel("Instructor")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.histplot(
    instructor_performance[
        "InstructorPerformanceScore"
    ],
    bins=10,
    kde=True
)

plt.title("Instructor Performance Score Distribution")
plt.xlabel("Performance Score")
plt.ylabel("Number of Instructors")

plt.show()
```

```python
expertise_analysis = (
    instructor_performance
    .groupby("Expertise")
    .agg(
        AverageTeacherRating=("TeacherRating", "mean"),
        AverageCourseRating=("AverageCourseRating", "mean"),
        AveragePerformanceScore=(
            "InstructorPerformanceScore",
            "mean"
        ),
        TotalEnrollments=("EnrollmentCount", "sum"),
        InstructorCount=("TeacherID", "nunique")
    )
    .sort_values(
        "AveragePerformanceScore",
        ascending=False
    )
)

display(expertise_analysis)
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    data=expertise_analysis.reset_index(),
    x="AveragePerformanceScore",
    y="Expertise"
)

plt.title("Average Instructor Performance by Expertise")
plt.xlabel("Average Performance Score")
plt.ylabel("Expertise")

plt.tight_layout()
plt.show()
```

```python
def experience_group(years):
    if years <= 5:
        return "0-5 Years"
    elif years <= 10:
        return "6-10 Years"
    elif years <= 20:
        return "11-20 Years"
    else:
        return "20+ Years"


instructor_performance["ExperienceGroup"] = (
    instructor_performance[
        "YearsOfExperience"
    ].apply(experience_group)
)

display(
    instructor_performance[
        [
            "TeacherName",
            "YearsOfExperience",
            "ExperienceGroup",
            "TeacherRating"
        ]
    ].head()
)
```

```python
experience_analysis = (
    instructor_performance
    .groupby("ExperienceGroup")
    .agg(
        AverageTeacherRating=("TeacherRating", "mean"),
        AverageCourseRating=("AverageCourseRating", "mean"),
        AveragePerformanceScore=(
            "InstructorPerformanceScore",
            "mean"
        ),
        InstructorCount=("TeacherID", "nunique")
    )
)

display(experience_analysis)
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    data=experience_analysis.reset_index(),
    x="ExperienceGroup",
    y="AveragePerformanceScore"
)

plt.title(
    "Instructor Performance by Experience Group"
)

plt.xlabel("Experience Group")
plt.ylabel("Average Performance Score")

plt.tight_layout()
plt.show()
```

```python
high_performer_enrollments = instructor_performance.loc[
    instructor_performance["PerformanceTier"] == "High Performer",
    "EnrollmentCount"
].sum()

total_enrollments = instructor_performance[
    "EnrollmentCount"
].sum()

enrollment_influence_ratio = (
    high_performer_enrollments /
    total_enrollments
) * 100

print(
    "Enrollment Influence Ratio:",
    round(enrollment_influence_ratio, 2),
    "%"
)
```

```python
experience_correlation = teachers[
    [
        "YearsOfExperience",
        "TeacherRating"
    ]
].corr().iloc[0, 1]

experience_impact_score = (
    abs(experience_correlation) * 100
)

print(
    "Experience Impact Score:",
    round(experience_impact_score, 2)
)
```

```python
rating_mean = teachers["TeacherRating"].mean()
rating_std = teachers["TeacherRating"].std()

rating_consistency_index = (
    1 - (rating_std / rating_mean)
) * 100

print(
    "Rating Consistency Index:",
    round(rating_consistency_index, 2)
)
```

```python
kpis = {
    "Average Teacher Rating":
        round(teachers["TeacherRating"].mean(), 2),

    "Average Course Rating":
        round(courses["CourseRating"].mean(), 2),

    "Rating Consistency Index":
        round(rating_consistency_index, 2),

    "Experience Impact Score":
        round(experience_impact_score, 2),

    "Enrollment Influence Ratio":
        round(enrollment_influence_ratio, 2),

    "Total Instructors":
        teachers["TeacherID"].nunique(),

    "Total Courses":
        courses["CourseID"].nunique(),

    "Total Enrollments":
        transactions["TransactionID"].nunique()
}

for key, value in kpis.items():
    print(f"{key}: {value}")
```
